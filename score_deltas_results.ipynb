```python
import pandas as pd
# Complex_Network_Analysis, Solar_Cells, Gravitational_Waves, Magnetic_Thin_Films, LHC
TOPIC_NAME = "LHC"
```

```python
df_deltas = pd.read_csv(f"data/score_deltas_by_year_{TOPIC_NAME}_2005_2025.csv")
df_deltas.columns = ["institution_id", "score_2025", "score_2005"]
geo_dict = pd.read_csv(f"data/geo_dict_{TOPIC_NAME}.csv")
geo_dict.columns = ["institution_id", 'name', 'country', 'latitude', 'longitude']

#use geo_dict to add institution name and country to df_deltas
df_deltas = df_deltas.merge(geo_dict[["institution_id", "name", "country"]], on="institution_id", how="left")
#only keep institutions that don't have the same sign for score_2025 and score_2005
df_deltas = df_deltas[df_deltas["score_2025"] * df_deltas["score_2005"] < 0]
#sort by absolute value of score_2025 - score_2005
df_deltas["score_delta"] = df_deltas["score_2025"] - df_deltas["score_2005"]
df_deltas = df_deltas.sort_values(by="score_delta", key=abs, ascending=False)
```

```python
# save name, country, score_2005, score_2025 to csv
df_deltas[["name", "country", "score_2005", "score_2025"]].to_csv(f"data/results/score_deltas_{TOPIC_NAME}.csv", index=False)
```
