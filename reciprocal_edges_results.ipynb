```python
import pandas as pd
# Complex_Network_Analysis, Solar_Cells, Gravitational_Waves, Magnetic_Thin_Films, LHC
TOPIC_NAME = "Complex_Network_Analysis"
```

```python
df_edges = pd.read_csv(f"data/reciprocal_edges_by_year_{TOPIC_NAME}_2005_2025.csv")
#first column is a tupe of (year, (institution1, institution2)), change it to three columns: year, institution1, institution2
df_edges[["year", "institution1", "institution2"]] = df_edges["Unnamed: 0"].str.extract(r"\((\d+), \('(.+)', '(.+)'\)\)")
df_edges = df_edges.drop(columns=["Unnamed: 0"])
#rename column 0 into weight
df_edges = df_edges.rename(columns={'0': "weight"})
geo_dict = pd.read_csv(f"data/geo_dict_{TOPIC_NAME}.csv")
geo_dict.columns = ["institution_id", 'name', 'country', 'latitude', 'longitude']
#use geo_dict to add insitution names and countries to df_edges
df_edges = df_edges.merge(geo_dict[["institution_id", "name", "country"]], left_on="institution1", right_on="institution_id", how="left").drop(columns=["institution_id"]).rename(columns={"name": "institution1_name", "country": "institution1_country"})
df_edges = df_edges.merge(geo_dict[["institution_id", "name", "country"]], left_on="institution2", right_on="institution_id", how="left").drop(columns=["institution_id"]).rename(columns={"name": "institution2_name", "country": "institution2_country"})
#open file to write results
with open(f"data/results/reciprocal_edges_{TOPIC_NAME}.txt", "w") as f:
    for year in range(2005, 2026):
        print(f"Year: {year}")
        f.write(f"Year: {year}\n")
        df_year = df_edges[df_edges["year"] == str(year)]
        #print institution1 name (country) <-> institution2 name (country) : weight
        for index, row in df_year.iterrows():
            print(f"{row['institution1_name']} ({row['institution1_country']}) <-> {row['institution2_name']} ({row['institution2_country']}): {int(row['weight'])}")
            f.write(f"{row['institution1_name']} ({row['institution1_country']}) <-> {row['institution2_name']} ({row['institution2_country']}): {int(row['weight'])}\n")
```
